# 02 · Two Pointers

> Due indici che si muovono sulla stessa sequenza, di solito dagli estremi verso il centro. Funziona quando ogni confronto ti dice **quale** indice spostare senza perdere soluzioni, tipicamente perché l'array è ordinato.
> Idee per i singoli problemi: [Pattern algoritmici › Two pointers](../Fundamentals/Pattern%20algoritmici.md#two-pointers)

## Problemi → cosa ti serve

| Problema | Concetti | Sintassi |
|---|---|---|
| Valid Palindrome | estremi verso il centro, salta i caratteri non validi | `ch.isalnum()`, `ch.lower()`, `while l < r` |
| Two Sum II | array ordinato: somma troppo piccola → `l += 1`, troppo grande → `r -= 1` | indici 1-based nel risultato: `[l + 1, r + 1]` |
| 3Sum | ordina, fissa `i`, two pointers su `i+1..n-1`; salta i duplicati | `nums.sort()`, `if i > 0 and nums[i] == nums[i - 1]: continue` |
| Container With Most Water | area = altezza minore × distanza; sposta il lato più basso | `min(a, b) * (r - l)`, `max(best, area)` |
| Trapping Rain Water | l'acqua su una cella dipende dal massimo a sinistra e a destra | `left_max`, `right_max` aggiornati mentre muovi i puntatori |

## Sintassi al volo

| Serve | Sintassi |
|---|---|
| ciclo con due indici | `l, r = 0, len(xs) - 1` · `while l < r:` |
| scambio | `xs[l], xs[r] = xs[r], xs[l]` |
| ordinare sul posto / copia | `xs.sort()` (restituisce `None`) · `sorted(xs)` |
| stringa rovesciata | `s[::-1]` · `reversed(s)` (iteratore) |
| filtrare caratteri | `"".join(ch.lower() for ch in s if ch.isalnum())` |
| aggiornare un massimo | `best = max(best, value)` |

## Estremi verso il centro

Lo scheletro base. Ogni iterazione sposta almeno un indice, quindi il ciclo fa al più `n` passi: `O(n)` tempo, `O(1)` spazio.

```python
def is_mirrored(code: str) -> bool:
    l, r = 0, len(code) - 1
    while l < r:
        if code[l] != code[r]:
            return False
        l += 1
        r -= 1
    return True

print(is_mirrored("AB-BA"), is_mirrored("AB-CA"))  # -> True False
```

## Saltare i caratteri da ignorare

Per ignorare spazi e punteggiatura senza costruire una nuova stringa, usa un `while` interno per ciascun puntatore. Ricorda la condizione `l < r` anche nei cicli interni, altrimenti esci dai limiti.

```python
coupon = "S-a, l e!"
print([ch for ch in coupon if ch.isalnum()])  # -> ['S', 'a', 'l', 'e']
print("A".lower() == "a", "7".isalnum(), "_".isalnum())  # -> True True False

l, r = 0, len(coupon) - 1
while l < r and not coupon[l].isalnum():
    l += 1
while l < r and not coupon[r].isalnum():
    r -= 1
print(coupon[l], coupon[r])  # -> S e
```

| Metodo | Vero per |
|---|---|
| `isalnum()` | lettere e cifre |
| `isalpha()` | solo lettere |
| `isdigit()` | solo cifre (`"-3".isdigit()` è `False`) |
| `isspace()` | spazi, tab, a capo |

## Su array ordinato: muovi in base al confronto

Se `xs` è ordinato e `xs[l] + xs[r]` è troppo piccolo, nessuna coppia con questo `l` può funzionare (tutti gli `r` più piccoli danno somme ancora minori): scarti `l`. Simmetrico per `r`.

```python
prices = sorted([450, 2599, 990, 1200, 300])  # -> [300, 450, 990, 1200, 2599]
budget = 1650

l, r = 0, len(prices) - 1
while l < r:
    total = prices[l] + prices[r]
    if total == budget:
        print(prices[l], prices[r])  # -> 450 1200
        break
    if total < budget:
        l += 1
    else:
        r -= 1
```

## Ordinare + fissare un elemento + saltare i duplicati

Per le triple (o k-uple), ordini una volta (`O(n log n)`), fissi il primo elemento con un `for` e cerchi gli altri con i due puntatori: `O(n²)` in totale.

Per non produrre la stessa tripla due volte, salti un valore uguale al precedente: sia per l'elemento fisso, sia per `l` dopo aver trovato una soluzione.

```python
nums = sorted([2, 2, 2, 5, 5, 7])  # -> [2, 2, 2, 5, 5, 7]

for i in range(len(nums)):
    if i > 0 and nums[i] == nums[i - 1]:
        continue  # stesso valore fisso di prima: risultati già trovati
    print("fisso", nums[i])  # -> fisso 2 / fisso 5 / fisso 7

l = 1  # hai appena trovato una soluzione con l = 0 e fatto l += 1
while l < r and nums[l] == nums[l - 1]:
    l += 1  # dopo una soluzione: avanza oltre i duplicati di nums[l]
print(l, nums[l])  # -> 3 5
```

## Massimi correnti da sinistra e da destra

In alcuni problemi ti serve per ogni posizione il massimo alla sua sinistra e alla sua destra. Due modi:

1. **Due array** (`O(n)` spazio): una passata in avanti per `left_max`, una all'indietro per `right_max`.
2. **Due puntatori** (`O(1)` spazio): sposti il lato con il massimo minore, perché è quello il collo di bottiglia e lo conosci già con certezza.

```python
heights = [0, 2, 1, 3, 1, 2]
n = len(heights)

left_max = [0] * n
for i in range(1, n):
    left_max[i] = max(left_max[i - 1], heights[i - 1])

right_max = [0] * n
for i in range(n - 2, -1, -1):
    right_max[i] = max(right_max[i + 1], heights[i + 1])

print(left_max)   # -> [0, 0, 2, 2, 3, 3]
print(right_max)  # -> [3, 3, 3, 2, 2, 0]
```

## Due puntatori nella stessa direzione

Variante "lettore/scrittore": `read` scorre tutto, `write` avanza solo quando tieni un elemento. Serve per compattare un array sul posto (rimuovere duplicati, spostare zeri).

```python
stock = [0, 3, 0, 0, 7, 1]
write = 0
for read in range(len(stock)):
    if stock[read] != 0:
        stock[write], stock[read] = stock[read], stock[write]
        write += 1
print(stock)  # -> [3, 7, 1, 0, 0, 0]
```

## Errori comuni

- `while l <= r` quando confronti due elementi distinti: con `l == r` useresti lo stesso elemento due volte.
- Dimenticare di spostare un puntatore in un ramo: ciclo infinito.
- 3Sum: saltare i duplicati **prima** di trovare la soluzione su `l`/`r` perde triple valide come `[-1, -1, 2]`.
- `xs.sort()` restituisce `None`: `xs = xs.sort()` cancella la lista.
